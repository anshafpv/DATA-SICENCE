```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv("/content/titanic.csv") #to read files
df
df.head() # to show first 5 rows
print(df.shape) #to show the no of rows and column
df.info() #it  shows column names, data types, and no
df.describe() #this gives a statistical summary of numarical mean,meadian,mode
df.isnull().sum()
df["Embarked"]=df["Embarked"].fillna(
    df["Embarked"].mode()[0]
)
print(df.isnull().sum())
print("Before:",df.shape[0])

df=df.drop_duplicates()

print("After:",df.shape[0])
df=df.rename(columns={
    "passengerid":"passengerID",
    "2urvived":"survived"
    })
df["Sex"]=df["Sex"].map({
    0:"Male",
    1:"Famale"
})
df
```

```python
sns.histplot(data=df,x="Age",bins=20)
plt.title("Age distribution of tatanic passengers")
plt.xlabel("Age")
plt.ylabel("number of passengers")
plt.show()
```

```python
sns.countplot(data=df,x="Sex",hue="survived")
plt.title("Titanic survival by sex")
plt.xlabel("sex")
plt.ylabel("number of passengers")
plt.show()
```

```python
sns.countplot(data=df,x="Pclass",hue="survived")
plt.title("Titanic survival by passenger class")
plt.xlabel("passenger class0")
plt.ylabel("numbers os passengers")
plt.show()
```
