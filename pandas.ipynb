```python
import pandas as pd
sales =pd.Series(
    [1200,850,1500,950,2000],
    index=["Monday","Tuesday","Wednesday","Thursday","Friday"]
)
print(sales)
print(sales.max())
print(sales.min())
print(sales.sum())
print(sales[sales<1000])
```

```python
import pandas as pd
data=[20,20,30,30,50,60,60,80]
ar=pd.Series(data)
print(ar)
print(ar.unique()) #avoid repeated value
print(ar.nunique()) #total cout
print(ar.value_counts()) #name and count
```
