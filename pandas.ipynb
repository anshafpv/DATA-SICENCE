```python
import pandas as pd
sales =pd.Series(
    [1200,850,1500,950,2000],
    index=["Monday","Tuesday","Wednesday","Thursday","Friday"]
)
print(sales)
print(sales.max())
print(sales.min())
print(sales.sum())
print(sales[sales<1000])
```
